#String Concatenation

```python
first_name="Sadman"
space=" "
last_name="Sakib"
full_name=first_name+space+last_name
print(full_name)
print(len(first_name))
print(len(space))
print(len(last_name))
print(len(full_name))
```

# Escape Sequence in String


```python
print("I am Sadman Sakib.\nI am 23 years old")   #adds a new line
print("Days\tTopic\tExercise")   #adds four space in place of \t
print("Day 1\t5\t5")
print("Day 2\t5\t6")
print("This is to print a backslash \\")#because if I do just \ this will accept a escape character sequence
print('\"')  #printing single/double quote in the same way
```

# String formatting (new style)

```python
first_name="Sadman"
last_name="Sakib"
age=10;
print("My name is {} {} and I am {} years old".format(first_name,last_name,age))

a=4;
b=3;
print ("{}+{}={}".format(a,b,a+b))
print("{}-{}={}".format(a,b,a-b))
print("{}*{}={}".format(a,b,a*b))
print("{}/{}={:.2f}".format(a,b,a/b))
print("{}//{}={}".format(a,b,a//b))
print("{}^{}={}".format(a,b,a**b))
radius=float(input("Enter radius:"))
pi=3.14
area=pi*radius*radius
print("The area of the circie of radius {} is : {:.3f}".format(radius,area))
```

# Accessing characters in string

```python
name="Sadman Sakib"
print(name[0])
last_index=len(name)-1;
print(name[last_index])
print(name[-1])   #negative indexing starts from backward
print(name[-2])
#print(name[12]) index out of bounds
```

# String Slicing

```python
line="python"
print(line[0:3]) #from 0 to 2 (end not inclusive)
print(line[3:7]) # 7 is out of bounds. Still it is valid as it will go till end
print(line[:])  #print the whole string
print(line[-3:6])
print(line[-3::-1]) # start from -3 index and go in backwards direction as the step is negative
print(line[::-1]) #reverses the string
print(line[0:6:2]) #skip elements. Print every two elements --->0,2,4
```
